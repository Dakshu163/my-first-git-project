```python
Q. 2 — Return More Than One Result
```

```python
def calculate_order(unit_price, quantity):
    subtotal = unit_price * quantity
    vat_amount = subtotal * 0.20
    final_total = subtotal + vat_amount
 
    return subtotal, vat_amount, final_total
 
 
subtotal, vat, total = calculate_order(25, 2)
 
print(subtotal)
print(vat)
print(total)
```

```python
Q. 3 — Reuse the Same Function for Three Orders
Order 1: price 10.00, quantity 4
Order 2: price 35.50, quantity 2
Order 3: price 7.25, quantity 10
```

```python
def calculate_order(unit_price, quantity):
    subtotal = unit_price * quantity
    vat_amount = subtotal * 0.20
    final_total = subtotal + vat_amount
    return final_total
 
 
print(f"Order 1: £{calculate_order(10.00, 4):.2f}")
print(f"Order 2: £{calculate_order(35.50, 2):.2f}")
print(f"Order 3: £{calculate_order(7.25, 10):.2f}")
```

```python
Q. 7 — Combine Two Functions
```

```python
def is_valid_amount(amount):
    return amount > 0
 
 
def apply_discount(amount, discount_percent=5):
    discount = amount * discount_percent / 100
    return amount - discount
 
 
amounts = [100, -10, 50]
 
for amount in amounts:
    if is_valid_amount(amount):
        print(apply_discount(amount, 10))
    else:
        print("Invalid amount")
```

```python
Q. 15 — Work with a Two-Dimensional Array
The rows are stores and the columns are three days. 
Calculate:
total sales for each store;
total sales for each day.
```

```python
import numpy as np
 
sales = np.array([
    [100, 120, 90],
    [80, 110, 105],
    [95, 100, 130],
])
 
store_totals = sales.sum(axis=1)
day_totals = sales.sum(axis=0)
 
print("Store totals:", store_totals)
print("Day totals:", day_totals)
```

```python
Q. 17 — Create a Reusable NumPy Summary Function
```

```python
import numpy as np
 
 
def summarise_sales(values):
    return {
        "total": values.sum(),
        "minimum": values.min(),
        "maximum": values.max(),
        "mean": values.mean(),
    }
 
 
sales = np.array([45.50, 18.00, 62.25, 30.00])
 
print(summarise_sales(sales))
```

```python
Q. 18 — Reject Invalid Values Before Aggregation
```

```python
import numpy as np
 
sales = np.array([45.50, -5.00, 18.00, 0, 62.25])
 
valid_sales = sales[sales > 0]
 
print("Valid values:", valid_sales)
print("Total:", valid_sales.sum())
print("Mean:", valid_sales.mean())
```

```python
Q. 20 — Analyse Three Store Batches
```

```python
import numpy as np
 
batch_1 = np.array([45.50, 18.00, 62.25])
batch_2 = np.array([30.00, 55.50, 12.00])
batch_3 = np.array([80.00, 20.00, 100.00])
 
all_sales = np.concatenate([
    batch_1,
    batch_2,
    batch_3,
])
 
average = all_sales.mean()
above_average = all_sales[all_sales > average]
 
print("Total:", all_sales.sum())
print("Average:", average)
print("Above average:", above_average)
print("Count:", above_average.size)
```
