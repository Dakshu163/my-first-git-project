```python
Q. 14 — Calculate Storage for Three Files
Use: files_mb = [100, 250, 400]
replication_factor = 3
Ans: 
```

```python
files_mb = [100, 250, 400]
replication_factor = 3
 
original_total = sum(files_mb)
 
replicated_total = (
    original_total * replication_factor
)
 
print("Original:", original_total, "MB")
print("Replicated:", replicated_total, "MB")
```
